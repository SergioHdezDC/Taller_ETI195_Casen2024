### 5.1
En este caso revisaremos la distribucion de las siguientes variables demograficas: Sexo ('sexo'), Pertenencia a un pueblo indigena ('r3'), Orientacion Sexual ('os1'). Consideramos estas variables como importantes a revisar ya que por lo general la cantidad de datos de ambos sexos deberia ser parecida y una diferencia grande entre estos puede provocar sesgos a la hora de analizar los datos. Por otro lado queremos revisar si existe una buena cantidad de datos de gente Indigena y de alguna Orientacion Sexual que no sea heterosexual, ya que normalmente son minorias y es necesario revisar si existe falta de representacion.

```python
# Conteos
vc_sexo = df_casen.value_counts('sexo')
vc_indigena = df_casen.value_counts('r3')
vc_orientacion = df_casen.value_counts('os1')

print('Sexo')
print(vc_sexo)
print('\n')

print('Indigena')
print(vc_indigena)
print('\n')

print('Orientacion Sexual')
print(vc_orientacion)
print('\n')
```

```python
# Graficos
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
total_filas = len(df_casen)

# 1. Gráfico por Sexo
plt.figure(figsize=(8, 5))
ax1 = sns.countplot(data=df_casen, x="sexo", hue="sexo", palette="Blues_r", legend=False)
plt.title("Distribución por Sexo", fontsize=12, fontweight="bold", pad=12)
plt.xlabel("Categoría", fontweight="bold")
plt.ylabel("Cantidad de personas", fontweight="bold")

# Añadir porcentajes arriba de cada barra
for p in ax1.patches:
    height = p.get_height()
    if height > 0:
        porcentaje = (height / total_filas) * 100
        ax1.annotate(
            f"{porcentaje:.2f}%",
            (p.get_x() + p.get_width() / 2.0, height),
            ha="center",
            va="bottom",
            fontsize=10,
            fontweight="bold",
            xytext=(0, 4),
            textcoords="offset points",
        )

# Ajustar límite Y para dar espacio a los textos
ax1.set_ylim(0, ax1.get_ylim()[1] * 1.15)
plt.tight_layout()
plt.show()


# 2. Gráfico por Pertenencia a Pueblos Indígenas (r3)
plt.figure(figsize=(8, 5))
ax2 = sns.countplot(data=df_casen, x="r3", hue="r3", palette="Greens_r", legend=False)
plt.title("Distribución por Pertenencia a Pueblos Indígenas", fontsize=12, fontweight="bold", pad=12)
plt.xlabel("Categoría", fontweight="bold")
plt.ylabel("Cantidad de personas", fontweight="bold")

for p in ax2.patches:
    height = p.get_height()
    if height > 0:
        porcentaje = (height / total_filas) * 100
        ax2.annotate(
            f"{porcentaje:.2f}%",
            (p.get_x() + p.get_width() / 2.0, height),
            ha="center",
            va="bottom",
            fontsize=10,
            fontweight="bold",
            xytext=(0, 4),
            textcoords="offset points",
        )

ax2.set_ylim(0, ax2.get_ylim()[1] * 1.15)
plt.tight_layout()
plt.show()


# 3. Gráfico por Orientación Sexual (os1)
plt.figure(figsize=(10, 5))
ax3 = sns.countplot(data=df_casen, x="os1", hue="os1", palette="Purples_r", legend=False)
plt.title("Distribución por Orientación Sexual", fontsize=12, fontweight="bold", pad=12)
plt.xlabel("Categoría", fontweight="bold")
plt.ylabel("Cantidad de personas", fontweight="bold")
plt.xticks(rotation=20, ha="right")

for p in ax3.patches:
    height = p.get_height()
    if height > 0:
        porcentaje = (height / total_filas) * 100
        ax3.annotate(
            f"{porcentaje:.2f}%",
            (p.get_x() + p.get_width() / 2.0, height),
            ha="center",
            va="bottom",
            fontsize=9,
            fontweight="bold",
            xytext=(0, 4),
            textcoords="offset points",
        )

ax3.set_ylim(0, ax3.get_ylim()[1] * 1.15)
plt.tight_layout()
plt.show()
```

Los graficos demuestran que hay un claro desbalance en la cantidad de datos. En cuanto a Sexo, la cantidad de datos que se tienen es bastante pareja cercana al 50%, pero es posible ver que cantidad de gente perteneciente a una poblacion indigena es muy poca comparada con los que no. Aunque el desbalance es mucho mayor en la cantidad de datos por Orientacion Sexual, ya que las personas Heterosexuales representan un 97.58% del total de los datos, siendo escasos los datos de las demas orientaciones sexuales

### 5.2
Algunas de las subpoblaciones interseccionales que existen serian los distintos sexos en cada orientacion (hombre heterosexual, mujer lesbiana, hombre bisexual, etc.), las cuales se pueden ver en el siguiente grafico

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Configuración del estilo visual
sns.set_theme(style="whitegrid")
plt.figure(figsize=(12, 6))

# 1. Crear tabla cruzada en porcentajes respecto al total
subpoblaciones_pct = (
    pd.crosstab(df_casen["os1"], df_casen["sexo"], normalize="all") * 100
).reset_index()

# Transformar datos a formato largo para Seaborn
df_plot = pd.melt(
    subpoblaciones_pct,
    id_vars=["os1"],
    value_vars=["Hombre", "Mujer"],
    var_name="Sexo",
    value_name="Porcentaje",
)

# 2. Crear el gráfico de barras dobles
ax = sns.barplot(
    data=df_plot,
    x="os1",
    y="Porcentaje",
    hue="Sexo",
    palette=["#4C72B0", "#DD8452"],
)

plt.title(
    "Intersección: Orientación Sexual por Género (%)",
    fontsize=13,
    fontweight="bold",
    pad=15,
)
plt.xlabel("Orientación Sexual (os1)", fontweight="bold")
plt.ylabel("Porcentaje del total de la muestra (%)", fontweight="bold")
plt.xticks(rotation=15)
plt.legend(title="Sexo")

# 3. Agregar porcentajes sobre cada barra
for p in ax.patches:
    height = p.get_height()
    if height > 0:
        ax.annotate(
            f"{height:.2f}%",
            (p.get_x() + p.get_width() / 2.0, height),
            ha="center",
            va="bottom",
            fontsize=9,
            fontweight="bold",
            xytext=(0, 3),
            textcoords="offset points",
        )

plt.tight_layout()
plt.show()
```

Es claramente visible que la representacion de las orientaciones distintas a la heterosexual es casi nula. Al tener tan poca representacion es posible ver que al ver como se distribuyen otros atributos en cada subpoblacion aparezcan sesgos como el mostrado en el siguiente grafico

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Configuración de estilo visual
sns.set_theme(style="whitegrid")
plt.figure(figsize=(12, 6))

# 1. Crear la variable interseccional de Sexo y Orientación Sexual
df_casen["subpoblacion_os"] = df_casen["sexo"] + " - " + df_casen["os1"]

# 2. Calcular el % de analfabetismo dentro de cada subpoblación
# Reemplaza 'No, ninguno' si en tu dataset el valor numérico es 2 o dice 'No'
df_analf_os = (
    df_casen.groupby("subpoblacion_os")["e1"]
    .apply(lambda x: (x == "No, ninguno").mean() * 100)
    .reset_index()
)
df_analf_os.columns = ["Subpoblación", "Porcentaje_Analfabeto"]

# 3. Generar el gráfico de barras
ax = sns.barplot(
    data=df_analf_os,
    x="Subpoblación",
    y="Porcentaje_Analfabeto",
    hue="Subpoblación",
    palette="Reds_r",
    legend=False,
)

# Títulos y formato del gráfico
plt.title(
    "Porcentaje de Analfabetismo por Género y Orientación Sexual",
    fontsize=13,
    fontweight="bold",
    pad=15,
)
plt.xlabel("Subpoblación Interseccional (Sexo - Orientación)", fontweight="bold")
plt.ylabel("% de Analfabetismo dentro del Grupo", fontweight="bold")
plt.xticks(rotation=45, ha="right")

# Escalar el eje Y para que los textos superiores no se corten
max_val = df_analf_os["Porcentaje_Analfabeto"].max()
plt.ylim(0, max_val + (max_val * 0.25) if max_val > 0 else 10)

# 4. Agregar la etiqueta del porcentaje exacto sobre cada barra
for p in ax.patches:
    height = p.get_height()
    if height >= 0:
        ax.annotate(
            f"{height:.2f}%",
            (p.get_x() + p.get_width() / 2.0, height),
            ha="center",
            va="bottom",
            fontsize=9,
            fontweight="bold",
            xytext=(0, 4),
            textcoords="offset points",
        )

plt.tight_layout()
plt.show()
```

Guiandose unicamente por el grafico se puede decir que los Hombres que dicen no saber su orientacion sexual tienen un mayor porcentaje de analfabetismo comparado a las demas subpolaciones, o que las mujeres bisexuales, lesbianas o pertenecientes a otra orientacion no son analfabetas. Podriamos intentar buscar una correlacion entre estos dos atributos, pero nosotros sabemos que se genera al haber poca representacion de estas subpoblaciones en los datos

**1. Existencia de desbalance:**
Al analizar las variables demográficas en el dataset de la encuesta CASEN 2024, identificamos un desbalance de clases, particularmente en la orientacion sexual. Aproximadamente, el 98% de la muesta se identifica como "Heterosexual", mientras que menos de un 3% dice pertenecer a otra orientacion. Este desbalance es un reflejo de la compodición demográfica del país.

**2. Consecuencias al entrenar modelos de Machine Learning:**
Si utilizamos estos datos sin un tratamiento previo para entrenar un modelo de Machine Learning, nos veremos enfrentados a la "paradoja de la exactitud" (Accuracy Paradox). El algoritmo se enfocará en la clase mayoritaria porque ese grupo minimiza el error global rápidamente. Debido a esto, el modelo tendrá un rendimiento estadístico deficiente para la clase de menor población. Si este modelo se usara para asignar subsidios o evaluar riesgo crediticio, generaría una gran tasa de falsos negativos mucho más alta para las orientaciones minorias, excluyéndolos sistemáticamente y automatizando la discriminación.

**3. Artículos de referencia:**

Para este análisis, nos basamos en la siguiente literatura académica:

*   **Impacto de la falta de representatividad:** En el artículo *"Gender Shades: Intersectional Accuracy Disparities in Commercial Gender Classification"* (Buolamwini & Gebru, 2018), las autoras demuestran cómo los datos de entrenamiento sesgados hacia rostros masculinos y de piel clara generaban tasas de error muy altas para mujeres de piel oscura. Esto comprueba que la falta de representación de minorías demográficas en los datos de origen se puede traducir directamente en sistemas que discriminan en la vida real a esos grupos.

*   **Consecuencias de datos desbalanceados e históricos:** En *"Big Data's Disparate Impact"* (Barocas & Selbst, 2016), los autores argumentan que los algoritmos de Machine Learning actúan como un espejo de la sociedad donde si un grupo demográfico es minoritario o sus datos reflejan desventajas previas, el modelo de *Data Mining*, además de heredar esa desigualdad,le dará una "neutralidad matemática", justificando decisiones bajo el nombre de la objetividad algorítmica.
