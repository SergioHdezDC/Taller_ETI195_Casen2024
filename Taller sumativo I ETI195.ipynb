    *  **Variables categóricas registradas como números (`int64` / `float64`):** La variable cualitativa `o1` (Condición de trabajo la semana pasada) se cargó como float.
```python
# Corrección de variables float a int
columnas_de_conteo = ["v27a", "v27b", "v29b"]
df_casen[columnas_de_conteo] = df_casen[columnas_de_conteo].astype(int)
# Correción variable float a category
df_casen["o1"] = df_casen["o1"].astype(int).astype("category")

columnas_cat_num = [
    'p2', 'p3', 'p4', 'p9', 'ind_hacina', 'h7e', 'h7f', 
    'r14', 'r17a', 'r17b', 'r17c', 'v3', 'v5', 'v23'
]
df_casen[columnas_cat_num] = df_casen[columnas_cat_num].astype('category')
df_casen.info()
```
